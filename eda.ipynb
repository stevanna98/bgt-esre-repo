```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from ipywidgets import interact, FloatSlider

# 1. Define the simulation grid
# Reduced resolution slightly (from 100 to 50) for smoother real-time rendering
fc_vals = np.linspace(0.0, 1.0, 50)
dist_vals = np.linspace(0, 150, 50)
FC_GRID, DIST_GRID = np.meshgrid(fc_vals, dist_vals)

def plot_3d_economy(lam):
    """Function to generate the 3D plot for a specific lambda value."""
    # Compute the Economy Weight for the current lambda
    WEIGHT_GRID = FC_GRID * np.exp(-lam * DIST_GRID)
    
    # Initialize the 3D plot
    fig = plt.figure(figsize=(10, 6))
    ax = fig.add_subplot(111, projection='3d')
    
    # Plot the surface
    # We fix vmin and vmax so the colors mean the same thing across all lambda values
    surface = ax.plot_surface(
        FC_GRID, 
        DIST_GRID, 
        WEIGHT_GRID, 
        cmap='viridis', 
        edgecolor='none', 
        alpha=0.9,
        vmin=0.0,
        vmax=1.0 
    )
    
    # Format axes and titles
    ax.set_title(f'Bullmore-Sporns Economy Landscape (λ = {lam:.3f})', fontsize=14, pad=15)
    ax.set_xlabel('Functional Connectivity (FC)', labelpad=10)
    ax.set_ylabel('Anatomical Distance (mm)', labelpad=10)
    ax.set_zlabel('Economy Weight', labelpad=10)
    
    # Lock the Z-axis from 0 to 1 so the surface visibly drops when lambda increases
    ax.set_zlim(0, 1)
    
    # Adjust viewing angle
    ax.view_init(elev=30, azim=230)
    
    # Add colorbar
    cbar = fig.colorbar(surface, ax=ax, shrink=0.5, aspect=10, pad=0.1)
    cbar.set_label('Economy Weight')
    
    plt.tight_layout()
    plt.show()

# 2. Attach the interactive slider to the plot function
interact(
    plot_3d_economy, 
    lam=FloatSlider(
        value=0.020, 
        min=0.0005, 
        max=0.500, 
        step=0.005, 
        description='Lambda (λ):',
        continuous_update=False # Set to True if your computer renders it fast enough!
    )
)
```
